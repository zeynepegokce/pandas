## Array Ayırma (Split)

```python
x = np.array([1,2,3,10,10,3,2,1])
```

```python
np.split(x, [3,5])
#       neyi böl, şununcuya kadar böl(değişkenleri sayarken 0'dan başlarsın)
#            n sayı, n+ 1 parça
```

```python
l, m, n = np.split(x, [3,5])
print(l)
print(m)
n
```

```python
#iki boyutlu ayırma
```

```python
ş = np.arange(16).reshape(4,4)
ş
```

```python
np.vsplit(ş, [2])
```

```python
ust, alt = np.vsplit(ş, [2])
print(ust)
alt
```

```python
np.hsplit(ş, 2)
```

```python
sol, sağ = np.hsplit(ş, 2)
print(sol)
sağ
```

## Sıralama (Sorting)

```python
np.sort(v)
#küçükten büyüğe   #asıl veriyi değiştirmedi
```

```python
v.sort()
v             # asıl veriyi değiştirdi
```
